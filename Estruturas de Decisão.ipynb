<pre>
Dica para determinar se um ano é bissexto: 
- São bissextos todos os anos múltiplos de 400, p.ex.: 1600, 2000, 2400, 2800...
- São bissextos todos os múltiplos de 4, exceto se for múltiplo de 100 mas não de 400,

ex1: 2004 é múltiplo de 4, mas não é múltiplo de 100, então é bissexto.

```python
ano_bissexto = int(input("Informe o ano que você quer saber se é Bissexto?")) # Todo ano é um número inteiro
# A lógica do programa
if ano_bissexto % 400 == 0:
    print(f"O ANO DE {ano_bissexto} é Bissexto")
elif ano_bissexto % 4 == 0 and ano_bissexto % 100 != 0:
    print(f"O ANO DE {ano_bissexto} é Bissexto")
else:
    print(f"O ANO DE {ano_bissexto} não é Bissexto")
```
